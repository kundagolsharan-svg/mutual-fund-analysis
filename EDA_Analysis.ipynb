# EDA Analysis: Mutual Fund Dataset

This notebook was programmatically generated. It contains preprocessing steps, plots, and 10 key findings (each with a chart reference).

```python
import pandas as pd
import numpy as np
from pathlib import Path
ROOT = Path(r"C:\Users\shara\OneDrive\Desktop\mutual fund analysis")
RAW = ROOT / 'data' / 'raw'
OUT = ROOT / 'outputs' / 'plots'
OUT.mkdir(parents=True, exist_ok=True)
# Load raw CSVs
fm = pd.read_csv(RAW / '01_fund_master.csv')
nav = pd.read_csv(RAW / '02_nav_history.csv')
aum = pd.read_csv(RAW / '03_aum_by_fund_house.csv')
sip = pd.read_csv(RAW / '04_monthly_sip_inflows.csv')
cat_in = pd.read_csv(RAW / '05_category_inflows.csv')
folio = pd.read_csv(RAW / '06_industry_folio_count.csv')
scheme_perf = pd.read_csv(RAW / '07_scheme_performance.csv')
txn = pd.read_csv(RAW / '08_investor_transactions.csv')
phold = pd.read_csv(RAW / '09_portfolio_holdings.csv')
bench = pd.read_csv(RAW / '10_benchmark_indices.csv')
# Parse dates (coerce errors to NaT)
nav['date'] = pd.to_datetime(nav['date'], errors='coerce')
aum['date'] = pd.to_datetime(aum['date'], errors='coerce')
bench['date'] = pd.to_datetime(bench['date'], errors='coerce')
phold['portfolio_date'] = pd.to_datetime(phold['portfolio_date'], errors='coerce')
txn['transaction_date'] = pd.to_datetime(txn['transaction_date'], errors='coerce')
sip['month'] = pd.to_datetime(sip['month'], format='%Y-%m', errors='coerce')
cat_in['month'] = pd.to_datetime(cat_in['month'], format='%Y-%m', errors='coerce')
folio['month'] = pd.to_datetime(folio['month'], format='%Y-%m', errors='coerce')
# Numeric conversions
nav['nav'] = pd.to_numeric(nav['nav'], errors='coerce')
phold['weight_pct'] = pd.to_numeric(phold.get('weight_pct', pd.Series()), errors='coerce')
phold['market_value_cr'] = pd.to_numeric(phold.get('market_value_cr', pd.Series()), errors='coerce')
bench['close_value'] = pd.to_numeric(bench.get('close_value', pd.Series()), errors='coerce')
aum['aum_crore'] = pd.to_numeric(aum.get('aum_crore', pd.Series()), errors='coerce')
sip['sip_inflow_crore'] = pd.to_numeric(sip.get('sip_inflow_crore', pd.Series()), errors='coerce')
cat_in['net_inflow_crore'] = pd.to_numeric(cat_in.get('net_inflow_crore', pd.Series()), errors='coerce')
print('Data loaded: fund_master rows=', len(fm), 'nav rows=', len(nav))
```

# 10 Key Findings
1. The NAV series for the 40 schemes shows a pronounced upward trend during 2023 (labelled the '2023 Bull Run').
2. Multiple schemes experienced notable corrections in 2024 as highlighted in the NAV trends chart.
3. SBI is the dominant fund house in AUM among peers (see SBI highlight AUM chart).
4. Monthly SIP inflows steadily increased from 2022 and peaked at 31,002 Cr in Dec 2025 (annotated in SIP trend).
5. Category inflows reveal that Liquid and Sectoral/Thematic categories received large net inflows in several months (heatmap).
6. Investor base skews toward middle age groups — see age distribution pie chart.
7. Transaction amounts vary widely across age groups (box plot shows wide IQRs and outliers).
8. Geographic concentration: top few states account for the bulk of transaction amount (amount_by_state chart).
9. Folio counts nearly doubled between Jan 2022 (13.26 Cr) and Dec 2025 (26.12 Cr), marking strong industry expansion.
10. Pairwise correlations of daily returns (selected funds) show groups of highly correlated funds suggesting common factor exposures (correlation heatmap).
